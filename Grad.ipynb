```python
x=torch.tensor(4.0)
y=torch.tensor(0.0)

w=torch.rand(1,requires_grad=True)
b=torch.zeros(1,requires_grad=True)
```

```python
def cross_entropy_loss(prediction,true_value):
  eplison=1e-7
  prediction=torch.clamp(prediction,eplison,1-eplison)
  return -((true_value*torch.log(prediction))+((1-true_value)*torch.log(1-prediction)))
```

```python
z=x*w+b
```

```python
prediction=torch.sigmoid(z)
prediction
```

```python
loss=cross_entropy_loss(prediction,y)
loss
```

```python
loss.backward()
```

```python
w.grad
```

```python
b.grad
```
